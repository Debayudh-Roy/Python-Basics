# Importing Modules in Python: Modules and Packages

## In Python, modules and packages help organize and reuse code. Here's a comprehensive guide on how to import the packages and use their modules/functions.

```python
import math              ## math is a package here
val=math.sqrt(16)        ## sqrt is a module or function inside the package
print(val)
```

#### another way to accessthe modules/packages

```python
from math import sqrt,pi
print(sqrt(16))
print(pi)
```

```python
from math import *  ##It will import all the modules inside the math package
print(sqrt(25))
```

## Creating our own package

```python
from package.maths import addition
val=addition(2,3)
print(val)
```

## Accessing subpackage inside a package

```python
from package.subpackage import mult

result=mult.multiplication(4,5)
print(result)
```
