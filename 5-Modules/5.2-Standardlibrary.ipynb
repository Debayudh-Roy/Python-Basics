#### working on any csv file

```python
import csv

with open('example.csv', mode='w', newline='') as file:
    writer=csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['Debayudh','23'])

with open('example.csv', mode='r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

#### Date and Time

```python
from datetime import datetime,timedelta

now=datetime.now()
print(now)

yesterday=now-timedelta(days=1)
print(yesterday)
```

```python
import time
print(time.time())
time.sleep(2)
print(time.time())
```

#### Regular Expression

```python
import re

pattern=r'\d+'      ### It will find a group of digits
text='There are 123 apples and 456 bananas'
match=re.search(pattern,text)
print(match.group())    ## As it has already found 123 thats why not moving further
```
