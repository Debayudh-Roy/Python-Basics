### SQL and SQLite

#### SQL(Structured Query Language) is a standard language for managing and manipulating relational databases. SQLite is a self-contained, serverless, and zero configuration database engine that is widely used for embedded database systems.

```python
import sqlite3
```

```python
### Connect to a SQLite database

connection = sqlite3.connect('example.db')
connection
```

```python
cursor = connection.cursor()
```

```python
### Creating a Table

cursor.execute('''
    CREATE TABLE IF NOT EXISTS EMPLOYEES(
        ID INTEGER PRIMARY KEY,
        NAME TEXT NOT NULL,
        AGE INTEGER,
        DEPARTMENT TEXT
    )
''')


### Save the changes permanently to the database

connection.commit()
```

```python
cursor.execute('''
    SELECT * FROM EMPLOYEES
''')
```

```python
cursor.execute('''
    INSERT INTO EMPLOYEES(NAME, AGE, DEPARTMENT)
    VALUES('Debayudh',23,'PAT')
''')

cursor.execute('''
    INSERT INTO EMPLOYEES(NAME, AGE, DEPARTMENT)
    VALUES('Rahul',28,'Associate')
''')

cursor.execute('''
    INSERT INTO EMPLOYEES(NAME, AGE, DEPARTMENT)
    VALUES('John',30,'TL')
''')


connection.commit()
```

```python
### Query to get the data from the table

cursor.execute('SELECT * FROM EMPLOYEES')
rows=cursor.fetchall()

for row in rows:
    print(row)
```

```python
### Update the data in the table

cursor.execute('''
    UPDATE EMPLOYEES
    SET DEPARTMENT='PA'
    WHERE NAME='Debayudh'
''')

connection.commit()
```

```python
cursor.execute('SELECT * FROM EMPLOYEES')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
### Delete data from the table

cursor.execute('''
    DELETE FROM EMPLOYEES
    WHERE NAME = 'John'
''')

connection.commit()
```

```python
cursor.execute('SELECT * FROM EMPLOYEES')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
### Working with Sales Data

#Connect with a SQLite Database

connection = sqlite3.connect('sales_data.db')
cursor = connection.cursor()

### create a table for Sales data

cursor.execute('''
    CREATE TABLE IF NOT EXISTS SALES(
        ID INTEGER PRIMARY KEY,
        DATE TEXT NOT NULL,
        PRODUCT TEXT NOT NULL,
        SALES INTEGER,
        REGION TEXT
    )
''')

### Insert data into the sales table

sales_data = [
    ('2023-01-01', 'Product1', 100, 'North'),
    ('2023-01-02', 'Product2', 200, 'South'),
    ('2023-01-03', 'Product1', 150, 'East'),
    ('2023-01-04', 'Product3', 250, 'West'),
    ('2023-01-05', 'Product2', 300, 'North')
]

cursor.executemany('''
    INSERT INTO SALES(DATE,PRODUCT,SALES,REGION)
    VALUES(?,?,?,?)
''',sales_data)

connection.commit()
```

```python
cursor.execute('SELECT * FROM SALES')
rows=cursor.fetchall()

for row in rows:
    print(row)
```

```python
### close the connection

connection.close()
```
