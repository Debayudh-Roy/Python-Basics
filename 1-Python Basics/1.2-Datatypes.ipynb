##Data Types are classification of data which tell compiler or interpreter how the programmer intends to use the data.

```python
#Integer Example
age=23
print(type(age))
```

```python
#Floating Point Datatype
height = 5.7
print(type(height))
```

```python
#String Datatype
name="Debayudh"
print(type(name))
```

```python
#Boolean Datatype
is_true=True
print(type(is_true))
```
