##Arithmetic Operators

```python
a=10
b=5

add_result=a+b #addition
sub_result=a-b #subtraction
mult_result=a*b #multiplication
div_result=a/b #Division
floor_dov_result=a//b #floor division
modulus_result=a%b #modulus operation
exponent_opeation=a**b #Exponentiation

print(add_result)
print(sub_result)
print(mult_result)
print(div_result)
print(floor_dov_result)
print(modulus_result)
print(exponent_opeation)
```
