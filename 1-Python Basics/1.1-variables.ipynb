Variables are fundamental elements in programming used to store data that can be referenced and manipulated in a program. Python variables are created when we assign values to them, and they don't need any explicit declaration to reserve the memory. It happens automatically when we assign values to them.

```python
## Declaring and Assiging Variables

age = 23
height = 5.7
name = "Debayudh"
is_Student = True

##Printing Variables

print("age: ", age)
print("height: ",height)
print("name: ", name)
print("Student: ", is_Student)
```

```python
##Understanding Variable Types
age=25 #Integer
height=5.7 #Float
name="Debayudh" #str
is_Student = True #Boolean

print(type(age))
print(type(height))
print(type(name))
print(type(is_Student))
```

```python
##Type cheching and Conversion

age=23
print(type(age))

age_str=str(age)
print(age_str)
print(type(age_str))
```

```python
##Dynamic typing
##Python allows the type of a variable to change as the program executes

var=10 #int
print("The value of var is:",var, type(var))

var="Hello"
print("The value of var is:",var, type(var))

var=3.14
print("The value of var is:",var, type(var))
```
