### Data Visualization using Seaborn

#### Seaborn is a Python visualization library based on Matplotlib that provides a high-level interface for drawing attractive and informative statistical graphics. Seaborn helps in creating complex visualizations with just a few lines of code.

```python
import seaborn as sns
```

```python
### Basic plotting with seaborn

tips=sns.load_dataset('tips')
tips
```

```python
import matplotlib.pyplot as plt
```

```python
### Create a scatter plot

sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title("Scatter Plot of Total Bill vs Tip")
plt.show()
```

```python
### Line Plot

sns.lineplot(x='size',y='total_bill',data=tips)
plt.title("Line plot for total Bill by size")
plt.show()
```

```python
### Categorical Plots
### BAR Plots

sns.barplot(x='day',y='total_bill',data=tips)
plt.title("Bar plot of Total Bill By Day")
plt.show()
```

```python
### Box Plot

sns.boxplot(x='day',y='total_bill',data=tips)
```

```python
### Violin Plot

sns.violinplot(x='day',y='total_bill',data=tips)
```

```python
### Histograms

sns.histplot(tips['total_bill'],bins=10,kde=True)
```

```python
#### KDE Plot

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
sns.kdeplot(tips['total_bill'],fill=True)

plt.subplot(1,2,2)
sns.kdeplot(tips['total_bill'],fill=False)
```

```python
### Pair Plot

sns.pairplot(tips)
```

```python
### Heatmap // Correlation

corr=tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
```

```python
import pandas as pd
```

```python
sales_df=pd.read_csv('data.csv')
sales_df.head()
```

```python
### Plot total_sales by product

sns.barplot(x='Product',y='Sales',data=sales_df,estimator=sum)
plt.title("Total Sales by product")
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.show()
```

```python
### Plot total Sales by Region

sns.barplot(x='Region',y='Sales',data=sales_df,estimator=sum)
plt.title("Total Sales by Region")
plt.xlabel('Region')
plt.ylabel('Sales')
plt.show()
```
