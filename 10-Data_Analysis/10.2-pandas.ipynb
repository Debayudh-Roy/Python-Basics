### Pandas-DataFrame And Series

#### Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable, and potentially heterogenous tabular data structure with labeled axes(rows and cloumns).

```python
import pandas as pd
```

```python
### Series
### Pandas Series is a one-dimensional array-like object that can hold any data type. It is similar to a column in a table.

data = [1,2,3,4,5]
series = pd.Series(data)
print("Series \n", series)
print(type(series))
```

```python
### Create a series from a dictionary
data = {
    'a':1,
    'b':2,
    'c':3
}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data=[10,20,40]
idx=['a','b','c']
pd.Series(data,index=idx)
```
