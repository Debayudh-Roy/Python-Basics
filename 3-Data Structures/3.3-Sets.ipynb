## Sets

sets are built-in data type in Python used to store collections of unique items. They are unordered, meaning that the elements do not follow a specific order, and they do not allow duplicate elements. Sets are useful for membership tests, eliminating duplicate entries, and performing mathematical set operations like union, intersection, difference and symmetric difference. 

```python
#create a set

my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set=set()
print(type(my_empty_set))
```

```python
# Converting a list into a set
my_set = set([1,2,3,4,5])
print(my_set)
```

```python
# Basic Set Operations
# Adding and Removing elements

my_set.add(6)
print(my_set)
```

```python
#Sets automatically ignores or removes duplicateb elements

my_set.add(7)  
print(my_set)
```

```python
# Remove elements from set

my_set.remove(3)
print(my_set)
```

```python
my_set.remove(10)
```

```python
my_set.discard(10)
print(my_set)
```

Remove will give an error if we try to remove an element which is not present in the set. In that scenarion discard will not give an error.

```python
# pop method
# It pops an arbitary element from the set

removed_element=my_set.pop()
print(removed_element)
print(my_set)
```

```python
## Set membership test

my_set = {1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)
```

Mathematical Operations

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

## Union Operation

union_set=set1.union(set2)
print(union_set)

## Intersection operation

intersection_set=set1.intersection(set2)
print(intersection_set)

## Intersection and updatind an exisitng set

set1.intersection_update(set2)
print(set1)
```

```python
#Difference

set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

diff_set1 = set1.difference(set2)
diff_set2 = set2.difference(set1)

print(diff_set1)
print(diff_set2)
```

 Symmetric difference --->  returns a new set containing all elements that are present in either of the sets, but not in both

```python
symmetric_diff_set=set1.symmetric_difference(set2)
print(symmetric_diff_set)
```

Sets Methods

```python
set1={1,2,3,4,5}
set2={3,4,5}

# is subset
print(set1.issubset(set2))
print(set2.issubset(set1))

# is superset
print(set1.issuperset(set2))
print(set2.issuperset(set1))
```

```python
### Counting unique words in a text

text = "In this tutorial we are discussing about sets and sets are very important."
words=text.split() ## This will convert the text into list of words
print("Total words count: ",len(words))

## Convert this list of words into set to get unique words

unique_words=set(words)
print(unique_words)
print("Total number of unique words are: ",len(unique_words))
```
