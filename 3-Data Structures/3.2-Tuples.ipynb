## Tuples

Tuples are ordered collections of items that are immutable.

```python
## creating a Tuple

empty_tuple=tuple()
print(empty_tuple)
print(type(empty_tuple))
```

```python
numbers=tuple([1,2,3,4,5])  #passing a list in a tuple
print(numbers)
```

```python
list((1,2,3,4,5)) #Converting a tuple into a list
```

```python
mixed_tuple=(1,"Hello World",3.14,True)
print(mixed_tuple)
```

```python
numbers=(1,2,3,4,5)
print(numbers[2])
print(numbers[-1])
```

```python
print(numbers[0:3])
print(numbers[::-1])
```

```python
concatenated_tuple=numbers+mixed_tuple
print(concatenated_tuple)
```

```python
print(mixed_tuple*3)
```

```python
# Immutable nature of tuple
# That means once elements of tuples are assigned they can't be changed

numbers = (1,2,3,4,5)
numbers[1]=10
```

Tuples Methods

```python
numbers_1 = (1,2,3,4,4,4,5,5,7)
print(numbers_1.count(4))
print(numbers_1.index(5)) ##returns the index of the 1st occurance
```

Packing and Unpacking of Tuples

```python
# Packing

packed_tuple=1,"Hello",3.14
print(packed_tuple)
```

```python
#Unpacking Tuple

a,b,c=packed_tuple

print(a)
print(b)
print(c)
```

```python
#Unpacking with *

numbers = (1,2,3,4,5,6)
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python
## Nested tuples

nested_tuple=((1,2,3),("a","b","c"),(True,False))
print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
# Nested tuple Nested list
lst=[[1,2,3,4],[6,7,8,9],(1,"Hello",3.14,"c")]
print(lst[2])
```

```python
#iterating over nested tuples

nested_tuple=((1,2,3),("a","b","c"),(True,False))

for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()
    
```
