## Dictionaries

Dictionaries are unordered collections of items. They store data in key-value pairs. Keys must be unique and immutable, while values can be of any type.

```python
## Creating dictionary
empty_dict={}
print(type(empty_dict))
```

```python
empty_dict=dict()
print(empty_dict)
```

```python
student={
    "name":"Debayudh",
    "age":23,
    "company":"Cognizant"
}
print(student)
print(type(student))
```

Whenever we will use the same key the last where it has used will be retrieved.

```python
## Single key is always used

student={
    "name":"Debayudh",
    "age":23,
    "name":"Cognizant"
}
print(student)
```

```python
## Accessing dictionary elements

student={
    "name":"Debayudh",
    "age":23,
    "company":"Cognizant"
}

print(student)
```

```python
## Accessing dictionary elements

print(student["age"])
print(student["name"])

## Other ways of accessing
print(student.get("name"))
print(student.get("age"))
print(student.get("company"))
```

```python
## Accessing keys which are not there

print(student.get("last_name"))
print(student.get("last_name","Not Available")) ##Specifying default value
```

```python
## Dictionary elements are mutable so we can modify them

student["company"]="Google" #Updating value for the key
print(student)

student["address"]="India" ## added a new key and value
print(student)
```

```python
##delete key and value pair

del student["address"]

print(student)
```

```python
## Dictionary Methods

keys=student.keys() ## get all the keys
print(keys)

values=student.values() ## get all the values
print(values)

items=student.items() ##get all key value pairs
print(items)
```

```python
## Shallow copy

student_copy=student  ## This is hard copy
print(student)
print(student_copy)
```

```python
## Now chaning values in student

student["name"] = "Debayudh Roy"
print(student)
print(student_copy)  ## We can see this has also been changed
```

```python
student_copy1 = student.copy()  ## shallow copy
print(student)
print(student_copy1)
```

```python
student["name"] = "Debayudh"
print(student)
print(student_copy1)
```

Iterating over dictionaries

```python
## Iterating Over Keys

for keys in student.keys():
    print(keys)
```

```python
## Iterating Over values

for values in student.values():
    print(values)
```

```python
## Iterating Over key and value pairs

for key,value in student.items():
    print(f"{key}:{value}")
```

Nested Dictionaries

```python
students={
    "student1":{
        "name":"Debayudh",
        "age":23,
    },
    "student2":{
        "name":"Rahul",
        "age":24
    }
}

print(students)
```

```python
## Access nested dictionary elements

print(students["student1"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
# Iterating over nested dictionaries

for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

Dictionary Comprehension

```python
squares={x:x**2 for x in range(5)}
print(squares)
```

Conditional Dictionary Comprehension

```python
evens={x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## Merge 2 dictionaries into one

dict1={"a":1, "b":2}
dict2={"b":3, "c":4}
merged_dict={**dict1, **dict2}
print(merged_dict)
```

Use a dictionary to count the frequency of elements in list

```python
numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number] += 1
    else:
         frequency[number] = 1

print(frequency)
```
