Python Conditional Statements

Python Conditional Statement mainly contains if, elif and else keywords

if: Executes a block of code only if the given condition is True.

elif (else if): Checks another condition if the previous if or elif condition is False. 
Note: elif block is not mandatory and there can be multiple elif blocks in a code

else: Executes a block of code when none of the if or elif conditions are True.

```python
age=23

if age<13:
    print("You are a child")
elif age<19:
    print("You are a teenager")
else:
    print("You are an adult")
```

Nested Conditional Statements

```python
##Checking whether a number is even or odd

num=int(input("Enter a number"))

if num>0:
    print("The number is positive")
    if num%2==0:
        print("The numer is even")
    else:
        print("The number is odd")
else:
    print("The number is zero or negative")
```

```python
##Checking whether is year is leap year or not

year=int(input("Enter the year"))

if (year%4==0 and year%100!=0) or (year%400==0):
    print("entered year is leap year")
else:
    print("Entered year is not leap year")
```

```python
##Simple calculator program

num1=float(input("Enter the first number: "))
num2=float(input("Enter the second number: "))
operation=input("Enter the operator (+, -, /, *): ")

if operation == '+':
    result = num1 + num2
elif operation == '-':
    result = num1 - num2
elif operation == '*':
    result = num1 * num2
elif operation == '/':
    if num2 != 0:
        result = num1/num2
    else:
        result = "Error! can't divide by zero"
else:
    result = "Invalid Operation"

print("Result: ", result)

```
