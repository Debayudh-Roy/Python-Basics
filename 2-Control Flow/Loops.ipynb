```python
## Strings

name="Debayudh"
for i in name:
    print(i) 
```

```python
#While
#The loop executes until and unless the specified condition is true

count=0

while count<5:
    print(count)
    count+=1
```

```python
## Loop control statements

# break
## The break statement exits the loop permanently

for i in range(5):
    if i==3:
        break
    print(i)
```

```python
## Continue

##The continue statement skips the current iteration and continues with the next

#This loop will skip the even numbers
for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
# pass

## the pass statement is a null operation. It does nothing

for i in range(5):
    if i==3:
        pass
    print(i)
```

```python
# Nested Loops: Loop inside a loop

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
## Sum of first n naturals numbers using for loop

n=10
sum=0
for i in range(n+1):
    sum=sum+i

print(sum)
```

```python
## Sum of first n naturals numbers using while loop

n=10
sum=0
count=0

while count<n+1:
    sum=sum+count
    count+=1

print(sum)
print(count)
```

```python
#Prime numbers between 1 and 100

for num in range (1,101):
    if num>1:
        for i in range(2, num):
            if num%i==0:
                break
        else:               ##Note: we can have an else block for a for loop
            print(num)
    
```
