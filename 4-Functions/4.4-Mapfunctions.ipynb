## map() Function

the map() function applies a given function to all items in an input list (or any other iterable) and returns a map object (an iterator). This particularly useful for transforming data in a list comprehensively.

```python
def square(x):
    return x*x

numbers=[1,2,3,4,5]

lst=list(map(square,numbers))   ### mapping has been done here
print(lst)
```

```python
## Lambda function with map

numbers1=[6,7,8,9]
new_lst=list(map(lambda x:x*x, numbers1))
print(new_lst)
```

```python
## Map multiple iterables

numbers1=[1,2,3]
numbers2=[4,5,6]

added_numbers = list(map(lambda x,y:x+y,numbers1,numbers2))
print(added_numbers)
```

```python
words=["Apple","Banana","Cherry"]
upper_words=list(map(str.upper,words))
print(upper_words)
```

```python
def get_name(person):
    return person["name"]

people=[
    {"name":"Debayudh","age":"23"},
    {"name":"Rahul","gge":"25"}
]

names=list(map(get_name,people))
print(names)
```
