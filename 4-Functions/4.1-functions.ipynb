## Functions

A function is a block of code that performs a specific task. Functions help in organizing code, reusing code and improving readability.

```python
### Function Definition

def even_odd(num):
    if num%2==0:
        print("The number is even")
    else:
        print("The number is odd")
```

```python
## Calling the function

even_odd(24)
```

```python
even_odd(21)
```

```python
## Fucntion with default parameters

def greet(name="Guest"):
    print(f"Hello {name}")

greet("Debayudh")
greet()
```

Variable length arguments

```python
## Positional arguments

def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,"debayudh")
```

```python
## keywords arguments

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="debayudh",age="23",country="India")
```

```python
def print_details(*args,**kwargs):
    for val in args:
        print(f"Positional argument :{val}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")


print_details(1,2,3,4,"Debayudh",name="debayudh",age="23",country="India")
```

```python
## Return statement

def multiply(a,b):
    return a*b

result=multiply(2,3)
print(result)
```

```python
## Return multiple parameters

def multiply(a,b):
    return a*b,a

result1,result2=multiply(2,3)
print(result1,result2)
```
