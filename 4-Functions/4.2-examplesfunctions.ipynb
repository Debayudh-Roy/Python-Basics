Example 1: Temperature Conversion

```python
def convert_temperature(temp,unit):
    if unit=='C':
        return temp * 9/5 + 32  ## Celcius to fahrenheit
    elif unit=='F':
        return (temp-32)*5/9 ## Fahrenheit to celsius
    else:
        return None
    
print(convert_temperature(25,'C'))
print(convert_temperature(77,'F'))
```

Example 2: Password strength checker



```python
def is_strong_password(password):
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False
    else:
        return True

print(is_strong_password("WealPwd"))
print(is_strong_password("Str0ngPwd!"))
```

Example 3: calculate the total cost of item in a shopping cart

```python
def calculate_total_cost(cart):
    total_cost=0
    for item in cart:
        total_cost+=item['price'] * item['quantity']

    return total_cost

cart=[
    {'name':'Apple','price':20,'quantity':4},
    {'name':'banana','price':10,'quantity':5},
    {'name':'Orange','price':5,'quantity':8},
]

total_cost=calculate_total_cost(cart)
print(total_cost)
```

Example 4:Check if a string is palindrome

```python
def is_palindrome(s):
    s=s.lower().replace(" ","")
    return s==s[::-1]

print(is_palindrome("A man A plan a canal Panama"))
print(is_palindrome("Debayudh"))
```

Example 5: Calculate factorial using recursion

```python
def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n-1)

print(factorial(6))
```
