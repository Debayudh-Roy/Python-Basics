### Magic Methods

#### Magic methods in Python, also known as Dunder methods (double underscore methods), are special methods that start and end with double underscores. These methods enable to define the behavior of objects for built-in operations, such as arithmetic operations, comparisons, and more.

#### These methods are predefined in Python that we can override to change the behavior of our objects. Some common magic methods include:

```python
'''
__init__: Initializes a new instance of a class.
__str__: Returns a string representation of a object.
__repr__: Returns an official string representation of a object.
__len__: Returns the length of an object.
__getitem___: Gets an item from a container.
__setitem__: Sets an item to a container.
'''
```

```python
### Basic Methods 
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Object create for [{self.name},{self.age}]"

    def __repr__(self):
        return f"Person (name={self.name}, age={self.age})"


person1=Person("Rahul", 32)
print(person1)
print(repr(person1))
```
