### Encapsulation

#### Encapsulation is the concept of wrapping data (variables) and methods together as a single unit.It restricts direct access to some objects components, which means accidental interference and misuse of the data.

```python
### Encapsulation with Getter and Setter methods 
### public,protected,private variables or access modifiers

class Public_Person:
    def __init__(self,name,age):
        self.name = name   ### public variable
        self.age = age     ### public variable

def get_name(person):
    return person.name

person=Public_Person("Debayudh",23)
print(get_name(person)) 
```

```python
dir(person)
```

```python
class Private_Person:
    def __init__(self,name,age):
        self.__name = name    ### private variables
        self.__age = age      ### private variables


privatePerson=Private_Person("Debayudh",23)
print(get_name(privatePerson))
```

```python
dir(privatePerson)
```

```python
class Protected_Person:
    def __init__(self,name,age):
        self._name = name       ### Protected Variables
        self._age = age         ### Protected Variables

protectedPerson=Protected_Person("Debayudh",23)
print(get_name(protectedPerson))
```

```python
dir(protectedPerson)
```

```python
### Encapsulation with Getter and Setter
class Employee:
    def __init__(self,name,age):
        self.__name=name  ##Private access modifier or variable
        self.__age =age   ##Private Variable

    ### getter and setter method
    def get_name(self):
        return self.__name

    def get_age(self):
        return self.__age

    def set_name(self,name):
        self.__name = name
    
    def set_age(self,age):
        if age>0:
            self.__age = age
        else:
            print("Age cannot be negative.")

employee = Employee("Debayudh",23)
```

```python
print(employee.get_name())
print(employee.get_age())

employee.set_age(25)
print(employee.get_age())
```
