# Working With File Paths

### When working with files in Python, handling file paths correctly is crucial to ensure that our code works across different operating systems and environments. Python provides several modules and functions for working with file paths effectively.

```python
### Using the os module

import os
cwd=os.getcwd()
print(f"The current working directory is {cwd}")
```

```python
## create a new directory

new_directory="packge"
os.mkdir(new_directory)
print(f"Directory {new_directory} created")
```

```python
## Listing files and directories

items=os.listdir('.')   ### Here . denotes the current working directory
print(items)
for item in items:
    print(item)
```

```python
### Joining paths

dir_name="folder"
file_name="file.txt"
full_path=os.path.join(os.getcwd(),dir_name,file_name)
print(full_path)
```

##### Checking if a path of a directory or file exists or not

```python
path="example1.txt"

if os.path.exists(path):
    print(f"The path '{path}' exists")
else:
    print(f"The path '{path}' does not exists")
```

```python
path="example.txt"

if os.path.exists(path):
    print(f"The path '{path}' exists")
else:
    print(f"The path '{path}' does not exists")
```

#### Checking if a path is file or directory

```python
path="example.txt"

if os.path.isfile(path):
    print(f"Provided path '{path}' is a file")
elif os.path.isdir(path):
    print(f"Provided path '{path}' is a directory")
else:
    print(f"Provided path '{path}' is neither a directory nor a file")
```

#### Getting absolute path

```python
relative_path='example.txt'
absolute_path=os.path.abspath(relative_path)
print(absolute_path)
```
