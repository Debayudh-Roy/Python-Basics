## File Operation- Read and Write Files

### File handling is a crucial part of any programming language. Python provides built-in functions and methods to read from and write to files, both text and binary.

```python
### Read a whole file

with open('example.txt','r') as file:
    content=file.read()
    print(content)
```

```python
### Read a file line by line

with open('example.txt', 'r') as file:
    for line in file:
        print(line)
```

```python
with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip())
```

what is actually happening is for the first one every line is having a new line character('\n') at te end. That's why the gap between the lines were there.

```python
## Writing a file(Overwriting)

with open('example.txt','w') as file:
    file.write("Hello\n")
    file.write("I will go to Google thats my promise")
```

```python
### Write a file (without overwriting)
with open('example.txt','a') as file:
    file.write("\nI will work as hard as I have to for that")
```

```python
### Binary Files

## Writing a binary file
data = b'\x00\x01\x02\x03\x04'
with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
### Reading binary file
with open('example.bin','rb') as file:
    content=file.read()
    print(content)
```

### Read the content from a source text file and write to a destination text file

```python
# Cooying a file

with open('example.txt','r') as source_file:
    content=source_file.read()

with open('destination.txt','w') as destination_file:
    destination_file.write(content)
```

### Read a text file and count the number of lines, words and characters

```python
def count_text_file(file_path):
    with open(file_path,'r') as file:
        lines = file.readlines()      ## takes all the lines of the file as a list
        line_count=len(lines)
        word_count=sum(len(line.split()) for line in lines)
        char_count=sum(len(line) for line in lines)
    return line_count, word_count, char_count

file_path = 'Code_Check.txt'
lines, words, characters = count_text_file(file_path)
print(f"Lines: {lines}, Words: {words}, Characters: {characters}")
```

### The ---->  w+  <---- mode in python is used to open a file for both reading and writing. If the file does not exist, it will be created. If the file exists, its content is truncated(i.e., the file is overwritten)

```python
with open('example.txt','w+') as file:
    file.write("Google is my ambition\n")
    file.write("I will achieve it definitely \n")

    ## Now the cursor is at the last line.
    ## We have to move it to the first line to read the file from the beginning

    file.seek(0)

    ## Read the content

    content=file.read()
    print(content)
```
