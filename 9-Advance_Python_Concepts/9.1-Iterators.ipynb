### Iterators

Iterators are advanced Python concepts that allow for efficient looping and memory management. Iterators. Iterators provide a way to access elements of a collection sequentially without expsoing the underlying structure.

```python
my_list=[1,2,3,4,5,6]
for i in my_list:
    print(i)
```

```python
print(type(my_list))
print(my_list)
```

```python
### Iterator
iterator=iter(my_list)
print(type(iterator))
```

```python
iterator
```

```python
### Iterate through all the element

next(iterator)
```

```python
### As the iterator already iterated through the list we need to initialize it again

iterator = iter(my_list)
```

```python
try:
    print(next(iterator))
except StopIteration:
    print("There is no more items in the iterator")
```

```python
###String Iterator

my_string = "Hello"
string_iterator = iter(my_string)

print(next(string_iterator))
print(next(string_iterator))
```
