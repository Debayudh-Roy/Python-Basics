### Generators 

Generators are a simpler way to create iterators. They use the yield keyword to produce a series of values lazily, which means they generate values on the fly and do not store them in memory. 

```python
def square1(n):
    for i in range(3):
        return i**2
```

```python
square1(3)
```

```python
def square2(n):
    for i in range(3):
        yield i**2
```

```python
square2(3)
```

```python
for i in square2(3):
    print(i)
```

```python
#### similarly if we executed square1

for i in square1(3):
    print(i)
```

```python
a=square2(3)
print(a)
```

```python
try:
    print(next(a))
except StopIteration:
    print("No more values exists")
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

```python
gen=my_generator()
gen
```

```python
try:
    print(next(gen))
except StopIteration:
    print("No more values exists")
```

### Practical Example

Generators are particularly useful for reading large files because they allows us to process one line at a time without loading the entire file into memory.

```python
def read_large_file(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line
```

```python
file_path = "large_file.txt"

for line in read_large_file(file_path):
    print(line.strip())
```
