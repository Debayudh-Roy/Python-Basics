### Decorators

Decorators are a powerful and flexible feature in Python that allows us to modify the behavior of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code.

```python
### function copy
### closures
### decorators
```

```python
## fucntion copy
def welcome():
    return "Hello! Welcome"

welcome()
```

```python
wel=welcome
print(wel())
del welcome   ### Deleting the source function 
print(wel())
```

```python
##closures function

def main_welcome(msg):
    def sub_welcome():
        print("Hello welcome inside 1")
        print(msg)
        print("Hello welcome inside 2")
    return sub_welcome()
```

```python
main_welcome("Main Welcome")
```

```python
def main_welcome1(func,lst):
    def sub_welcome1():
        print("Hi")
        print(func(lst))
        print("Hello")
    return sub_welcome1()
```

```python
main_welcome1(len,[1,2,3])
```

```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening After the function is called.")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello")
```

```python
say_hello()
```

```python
## Decorators with argument

def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hii():
    print("Hello")
```

```python
say_hii()
```
